[60,70,80,55,68,77,55,88,96,46]

mean - gives average

mode - max frequency of the data


q1 =  sum of score / no of scores







```python
x= [50,70,70,90,45,76,99,88]

mean = sum(x)/len(x)
print("Mean is:", mean)
median = sorted(x)[len(x)//2]
print("Median is:", median)
mode = max(set(x), key = x.count)
print("Mode is:", mode)
```
