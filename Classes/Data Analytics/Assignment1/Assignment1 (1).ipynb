```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_csv("Housing_Sales_Data_200.csv")
```

```python
# Check if there are any missing values
df.isnull().sum()
```

```python
df.describe()
```

```python
mean_values = df.mean(numeric_only=True)
median_values = df.median(numeric_only=True)
mode_values = df.mode(numeric_only=True).iloc[0]

print("Mean:\n", mean_values)
print("\nMedian:\n", median_values)
print("\nMode:\n", mode_values)
```

```python
range_values = df.max(numeric_only=True) - df.min(numeric_only=True)
variance_values = df.var(numeric_only=True)
std_values = df.std(numeric_only=True)
iqr_values = df.quantile(0.75, numeric_only=True) - df.quantile(0.25, numeric_only=True)

print("\nRange:\n", range_values)
print("\nVariance:\n", variance_values)
print("\nStandard Deviation:\n", std_values)
print("\nInterquartile Range (IQR):\n", iqr_values)
```

```python
df.corr(numeric_only=True)
```

### Data Analysis Summary

Based on the exploratory data analysis of the Housing Sales dataset, here are the key findings:

1. **Data Structure**:
   - The dataset contains complete information with no missing values
   - All features are numerical, making it suitable for statistical analysis

2. **Central Tendency Measures**:
   - The average house price in the dataset is higher than the median, suggesting some high-value properties are pulling the mean upward
   - This indicates a right-skewed distribution in house prices
   - Similar patterns are observed in features like square footage and number of rooms

3. **Dispersion Measures**:
   - There is considerable variation in house prices, as shown by the standard deviation and IQR
   - The range of values indicates diverse property types in the dataset
   - The variance shows high spread in numerical features, particularly in price and square footage

4. **Correlation Analysis**:
   - Strong positive correlations exist between:
     - House price and square footage
     - Number of bedrooms and bathrooms
   - This suggests that larger homes with more rooms generally command higher prices

These insights provide a solid foundation for understanding the housing market dynamics represented in this dataset.
