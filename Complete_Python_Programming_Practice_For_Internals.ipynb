```python
# Q7: If a five-digit number is input through the keyboard, write a program to calculate the sum
# of its digits without using any loop. (Hint: Use the modulus operator ‘%’)
```

```python


number = int(input())
newnum = 0
if len(str(number))==5:
  while number > 0:
     newnum += number % 10     # get last digit
     number //= 10             # remove last digit

print(newnum)
```

Q9: Program to count the number of each vowel in a string.

```python



sent = input("Enter a string: ")
count = 0
vowel = ['a', 'e', 'i', 'o', 'u']

for i in sent.lower():
    if i in vowel:
        count += 1

print("Total vowels:", count)
```
