# NLP Pipeline for Merging Similar Questions on Quora

## Problem Statement
Design an end-to-end NLP system that detects semantically similar questions on Quora so that duplicate questions can be merged, improving answer quality, reducing redundancy, and enhancing user experience.

---

## 1. Data Acquisition

### Data Sources
1. **Historical Quora Data**
   - Previously merged and non-merged question pairs
   - Moderator-approved duplicate questions
   - User-reported duplicate flags

2. **Publicly Available Datasets**
   - Quora Question Pairs (Kaggle)
   - Stack Overflow duplicate question datasets

3. **User Interaction Signals**
   - Users clicking between similar questions
   - Shared or reused answers
   - Engagement metrics (views, upvotes)

### Labeling Strategy
- Binary labels: Duplicate / Not Duplicate
- Weak supervision using:
  - Answer overlap
  - Common tags
  - High semantic similarity scores

---

## 2. Text Preparation

### Text Cleaning
- Convert text to lowercase
- Remove HTML tags, URLs, and emails
- Remove special characters and extra whitespace
- Expand contractions (e.g., can't → cannot)
- Handle punctuation consistently

---

### Text Preprocessing
- Tokenization
- Stopword removal (selective, not aggressive)
- Lemmatization (preferred over stemming)
- Spelling normalization (US vs UK)
- Normalize numbers, dates, and units

---

### Advanced Text Preprocessing
Since the task is semantic similarity, advanced preprocessing is important:
- Synonym normalization (e.g., car ↔ automobile)
- Phrase detection (machine learning, data science)
- Question intent detection (how, why, what)
- Negation handling to preserve meaning
- Named Entity Recognition (NER)

---

## 3. Feature Engineering

### Lexical Features
- TF-IDF vectors
- N-grams (unigrams and bigrams)
- Jaccard similarity
- Word overlap ratio

---

### Semantic Features
- Sentence embeddings (BERT / Sentence-BERT)
- Cosine similarity between embeddings
- Topic similarity (LDA / BERTopic)
- Semantic similarity score

---

### Structural and Metadata Features
- Question length difference
- Tag overlap ratio
- Question type match
- Shared entities (NER overlap)
- Answer similarity (if answers exist)

---

## 4. Modelling

### Model Selection

#### Baseline Models
- Logistic Regression
- Support Vector Machine (SVM)
- XGBoost using handcrafted features

---

#### Production-Grade Model
- Siamese Neural Network with Sentence-BERT
  - Each question is encoded independently
  - Embeddings are compared using cosine similarity
  - Threshold-based classification for duplicates

Why Siamese Models:
- Efficient and scalable
- High semantic understanding
- Reusable embeddings for retrieval

---

## 5. Evaluation Metrics

### Intrinsic Evaluation Metrics
- Precision
- Recall
- F1-score
- ROC-AUC
- Accuracy at different similarity thresholds

Precision is prioritized to avoid incorrect merges.

---

### Extrinsic Evaluation Metrics
- Reduction in duplicate questions
- Improvement in answer quality
- User engagement metrics (CTR, dwell time)
- Moderator override rate
- Successful merge acceptance rate

---

## 6. Deployment Strategy

### System Architecture
1. Embedding Service
   - Precompute embeddings for all existing questions
2. Vector Search Engine
   - FAISS / Pinecone / Milvus
3. Real-Time Inference
   - Generate embedding for new question
   - Retrieve top-k similar questions
   - Apply similarity threshold
4. Human-in-the-loop
   - Manual review for low-confidence cases

---

## 7. Monitoring and Observability

### Monitoring Metrics
- False positive merge rate
- False negative rate
- Embedding drift
- Model latency and throughput
- Distribution shift in topics

---

### Alerting
- Sudden spike in incorrect merges
- Decrease in user engagement
- Significant change in similarity score distributions

---

## 8. Model Update Strategy

### Continuous Retraining
- Weekly or bi-weekly batch retraining
- Use newly merged questions and user feedback
- Incorporate moderator decisions

---

### A/B Testing
- Gradual rollout of new models
- Compare model performance against baseline
- Measure business and engagement metrics

---

### Active Learning
- Sample low-confidence predictions
- Send for human labeling
- Improves model performance with less data

---

## Final Summary
This system uses semantic embeddings with Siamese BERT architectures, combined with lexical and structural features, deployed using vector similarity search and continuous human feedback to accurately and safely merge duplicate questions on Quora.
